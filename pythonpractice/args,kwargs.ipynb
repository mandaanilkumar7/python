```python
def greet(*args):
    for name in args:
        print(f"Hello, {name}!")

```

```python
l=("Anil", "Ravi", "Meena")
greet(*l)
```

```python
def show_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")


```

```python
show_details(name="Anil", role="Data Engineer", location="Vemulawada")
```

```python
di={"name": "Anil", "role": "Data Engineer", "location": "Vemulawada"}
show_details(**di)
```
